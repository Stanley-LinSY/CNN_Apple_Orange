# 1)Data Splitting

```python
import os
import shutil

#------------------------------------------------------------------
#DATA SPLITTING

#DatasetV1 is our 400x400 images, it contains 3 directories Orange, Apple, Unknown.
orange_src = 'C:/Python/2120795_StanleyLin_Orange_2112422_WaiYanAung_Apple/DatasetV1/Orange/'
apple_src= 'C:/Python/2120795_StanleyLin_Orange_2112422_WaiYanAung_Apple/DatasetV1/Apple/'
unknown_src='C:/Python/2120795_StanleyLin_Orange_2112422_WaiYanAung_Apple/DatasetV1/Unknown/'

orange_data=os.listdir(orange_src)
apple_data=os.listdir(apple_src)
unknown_data=os.listdir(unknown_src)

from sklearn.model_selection import train_test_split

#DatasetV1's 3 directories Orange, Apple, Unknown is then distributed randomly into Test, Train, Valid. For example, Test file would
#contain the 3 directories Orange, Apple, Unknown. The same can be said for Train and Valid files, these are all contained in DatasetV2.

orange_train,orange_test = train_test_split(orange_data,test_size=0.05,random_state=1) #Trasnferring 5% of the data,and randomizing
orange_train,orange_validate=train_test_split(orange_train,test_size=0.05,random_state=1)

apple_train,apple_test=train_test_split(apple_data,test_size=0.05,random_state=1)
apple_train,apple_validate=train_test_split(apple_train,test_size=0.05,random_state=1)

unknown_train,unknown_test=train_test_split(unknown_data,test_size=0.05,random_state=1)
unknown_train,unknown_validate=train_test_split(unknown_train,test_size=0.05,random_state=1)

#After the random split, the images are sorted into DatasetV2 
for image in orange_train:
    shutil.copy(orange_src+image,'C:/Python/2120795_StanleyLin_Orange_2112422_WaiYanAung_Apple/DatasetV2/Train/Orange/')
for image in orange_test:
    shutil.copy(orange_src+image,'C:/Python/2120795_StanleyLin_Orange_2112422_WaiYanAung_Apple/DatasetV2/Test/Orange/')
for image in orange_validate:
    shutil.copy(orange_src+image,'C:/Python/2120795_StanleyLin_Orange_2112422_WaiYanAung_Apple/DatasetV2/Valid/Orange/')   

for image in apple_train:
    shutil.copy(apple_src+image,'C:/Python/2120795_StanleyLin_Orange_2112422_WaiYanAung_Apple/DatasetV2/Train/Apple/')
for image in apple_test:
    shutil.copy(apple_src+image,'C:/Python/2120795_StanleyLin_Orange_2112422_WaiYanAung_Apple/DatasetV2/Test/Apple/')
for image in apple_validate:
    shutil.copy(apple_src+image,'C:/Python/2120795_StanleyLin_Orange_2112422_WaiYanAung_Apple/DatasetV2/Valid/Apple/')  
    
for image in unknown_train:
    shutil.copy(unknown_src+image,'C:/Python/2120795_StanleyLin_Orange_2112422_WaiYanAung_Apple/DatasetV2/Train/Unknown/')
for image in unknown_test:
    shutil.copy(unknown_src+image,'C:/Python/2120795_StanleyLin_Orange_2112422_WaiYanAung_Apple/DatasetV2/Test/Unknown/')
for image in unknown_validate:
    shutil.copy(unknown_src+image,'C:/Python/2120795_StanleyLin_Orange_2112422_WaiYanAung_Apple/DatasetV2/Valid/Unknown/') 
```

# 4) Load Data

```python
import tensorflow as tf
from tensorflow.keras.layers import Input,Conv2D,Flatten,Dense,MaxPool2D,BatchNormalization
from tensorflow.keras.models import Model,Sequential
from PIL import Image
from PIL import ImageFilter 

# Define the size of the input images
IMAGE_SIZE=[224,224]
# Set the batch size for training
BATCH_SIZE=16
# Adam Optimizer with a learning rate of 1e-5
opt=tf.keras.optimizers.Adam(learning_rate=1e-5)

# Directory paths for training, testing, and validation data
train_dir ='C:/Python/2120795_StanleyLin_Orange_2112422_WaiYanAung_Apple/DatasetV3/Train/'
test_dir ='C:/Python/2120795_StanleyLin_Orange_2112422_WaiYanAung_Apple/DatasetV3/Test/'
valid_dir='C:/Python/2120795_StanleyLin_Orange_2112422_WaiYanAung_Apple/DatasetV3/Valid/'

# Creating a generator for training data
train_generator = train_datagen.flow_from_directory(train_dir,
                                                       target_size = (IMAGE_SIZE),
                                                       batch_size = BATCH_SIZE,
                                                       class_mode = 'categorical',
                                                       shuffle=True
                                                    # Shuffle the data for each epoch
                                                   )
    
# Creating a generator for testing data
test_generator = test_datagen.flow_from_directory(test_dir,
                                                       target_size = (IMAGE_SIZE),
                                                       batch_size = BATCH_SIZE,
                                                       class_mode = 'categorical',
                                                       shuffle=False
                                                  # Do not shuffle the data for testing
                                                 )

valid_generator = valid_datagen.flow_from_directory(valid_dir,
                                                       target_size = (IMAGE_SIZE),
                                                       batch_size = BATCH_SIZE,
                                                       class_mode = 'categorical',
                                                       shuffle=False
                                                   # Do not shuffle the data for validation
                                                   )

def scratchmodel(weights='imagenet',include_top=False):
    #Building model in sequential manner
    model=Sequential()
    
    #Convulotional layers
    model.add(Conv2D(filters=32,input_shape=IMAGE_SIZE+[3],kernel_size=(3,3),padding="same",activation="relu"))
    model.add(MaxPool2D(pool_size=(2,2),strides=(2,2)))
    
    model.add(Conv2D(filters=64,kernel_size=(3,3),padding="same",activation="relu"))
    model.add(MaxPool2D(pool_size=(2,2),strides=(2,2)))
    
    model.add(Conv2D(filters=128,kernel_size=(3,3),padding="same",activation="relu"))
    model.add(MaxPool2D(pool_size=(2,2),strides=(2,2)))
    
    model.add(Conv2D(filters=256,kernel_size=(3,3),padding="same",activation="relu"))
    model.add(MaxPool2D(pool_size=(2,2),strides=(2,2)))

    model.add(tf.keras.layers.BatchNormalization())
    return model

# Creating the model using the scratchmodel function
model=scratchmodel()

# Flattening the model output
x=tf.keras.layers.Flatten()(model.output)

# Adding dropout to reduce overfitting
x=tf.keras.layers.Dropout(0.6)(x)
x=tf.keras.layers.Dense(64, activation='relu',kernel_regularizer=tf.keras.regularizers.l2(1e-4))(x)
x=tf.keras.layers.Dropout(0.4)(x)
x=tf.keras.layers.Dense(32, activation='relu',kernel_regularizer=tf.keras.regularizers.l2(1e-4))(x)
x=tf.keras.layers.Dropout(0.2)(x)
x=tf.keras.layers.Dense(3,activation='softmax',kernel_regularizer=tf.keras.regularizers.l2(1e-4))(x)

# Creating the final model
model=Model(inputs=model.input, outputs=x)
# Compiling the model with the specified optimizer, loss, and metrics
model.compile(optimizer=opt, loss='categorical_crossentropy', metrics=['accuracy'])


#Create checkpoints and save the model on each completed epoch during training
checkpoint=tf.keras.callbacks.ModelCheckpoint(
    "C:/Python/2120795_StanleyLin_Pineapple_2112422_WaiYanAung_Apple/oau_model_400_e10.hdf5",
                                              monitor="val_loss",
                                              verbose=1,
                                              save_best_only=False,
                                              save_weights_only=False,
                                              mode="auto",
                                              save_req="epoch",\
                                              options=None,
                                            )

#Monitors the validation accuracy and stop based on its value
early=tf.keras.callbacks.EarlyStopping(monitor="val_accuracy",
                                       min_delta=0,
                                       patience=40,
                                       verbose=1,
                                       mode="auto",
                                      )

# Enable logging of device placement for debugging purposes
tf.debugging.set_log_device_placement(True)

try:
    # Specify the GPU device to be used (assuming GPU:0)
    with tf.device('/device:GPU:0'):
        # Training the model
        history = model.fit(
        train_generator,
        steps_per_epoch=len(train_generator),  
        epochs=10, #number of traning rounds
        validation_data = valid_generator,
        validation_steps=len(valid_generator),
        callbacks=[checkpoint,early]
        )
except RuntimeError as e:
  print(e)
```

# 6) Life detect

```python
w = 'http://172.22.95.100:4747/video'
import tensorflow as tf
import cv2
from keras.preprocessing import image
from PIL import Image, ImageOps
import numpy as np

def import_and_predict(image_data, model):
    size = (224, 224)    
    image = ImageOps.fit(image_data, size, Image.ANTIALIAS)
    image = image.convert('RGB')
    image = np.asarray(image)
    image = (image.astype(np.float32) / 255.0)

    img_reshape = image[np.newaxis,...]

    prediction = model.predict(img_reshape)
    prediction_label = np.argmax(prediction)
    percentage = np.max(prediction) * 100

    return prediction_label, percentage

model = tf.keras.models.load_model("C:/Python/2120795_StanleyLin_Orange_2112422_WaiYanAung_Apple/oau_model_400_e10.hdf5")

cap = cv2.VideoCapture(w)

if cap.isOpened():
    print("Camera OK")

while True:
    ret, original = cap.read()
    
    frame = cv2.resize(original, (224, 224))
    cv2.imwrite(filename='img.jpg', img=original)
    image = Image.open('img.jpg')

    prediction_label, percentage = import_and_predict(image, model)

    if prediction_label == 2:
        predict = "It is an Unknown!"
    elif prediction_label == 1:
        predict = "It is an Orange!"
    elif prediction_label == 0:
        predict = "It is an Apple!"
    
    predict_with_accuracy = f"{predict} Accuracy: {percentage:.2f}%"
    cv2.putText(original, predict_with_accuracy, (10, 30), cv2.FONT_HERSHEY_SIMPLEX, 0.9, (0, 255, 0), 2)
    cv2.imshow("Classification", original)
    
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break;
        
cap.release()
```
